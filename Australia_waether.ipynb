```python
#Memeriksa jumlah nilai yang hilang di setiap kolom
missing_values = train.isnull().sum()
missing_values[missing_values > 0]
```

```python
less = missing_values[missing_values < 87276].index
over = missing_values[missing_values >= 87276].index
```

```python
# Mengambil kolom ber-missing value rendah yang bertipe numerik
numeric_features = (
    train[less].select_dtypes(include=["int64", "float64"]).columns
)
train[numeric_features] = train[numeric_features].fillna(
    train[numeric_features].median()
)

# Mengambil kolom ber-missing value rendah yang bertipe kategori/object
categorical_features = train[less].select_dtypes(include=["object", "str"]).columns
for column in categorical_features:
  train[column] = train[column].fillna(train[column].mode()[0])
```

```python
# Menghapus kolom dengan terlalu banyak nilai yang hilang
df = train.drop(columns=over)
```

```python
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
```

```python
# Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1
```

```python
# Filter dataframe untuk hanya menyimpan baris yang tidak mengandung outliers pada kolom numerik
condition = ~((df[numeric_features] < (Q1 - 1.5 * IQR)) | (df[numeric_features] > (Q3 + 1.5 * IQR))).any(axis=1)
df_filtered_numeric = df.loc[condition, numeric_features]

# Menggabungkan kembali dengan kolom kategorikal
categorical_features = df.select_dtypes(include=['str']).columns
df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)
```

```python
import sklearn
```

```python
# Memisahkan fitur (X) dan target (y)
X = df_lencoder.drop(columns=['RainTomorrow'])  
y = df_lencoder['RainTomorrow']
```

```python
from sklearn.model_selection import train_test_split

# membagi dataset menjadi training dan testing 
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
# menghitung panjang/jumlah data 
print("Jumlah data: ",len(X))
# menghitung panjang/jumlah data pada x_train
print("Jumlah data latih: ",len(x_train))
# menghitung panjang/jumlah data pada x_test
print("Jumlah data test: ",len(x_test))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Model 1: Logistic Regression
logreg = LogisticRegression(random_state=1, max_iter=1000)
logreg.fit(x_train, y_train)

# Model 2: Decision Tree
dt = DecisionTreeClassifier(random_state=1)
dt.fit(x_train, y_train)

# Model 3: Random Forest
rf = RandomForestClassifier(random_state=1)
rf.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Fungsi helper untuk mempermudah evaluasi
def evaluate_model(model, name):
    pred = model.predict(x_test)
    return {
        'Accuracy': accuracy_score(y_test, pred),
        'Precision': precision_score(y_test, pred),
        'Recall': recall_score(y_test, pred),
        'F1-Score': f1_score(y_test, pred)
    }

# Evaluasi semua model ke dalam DataFrame
results = {
    'Logistic Regression': evaluate_model(logreg, 'Logistic Regression'),
    'Decision Tree': evaluate_model(dt, 'Decision Tree'),
    'Random Forest': evaluate_model(rf, 'Random Forest')
}

df_results = pd.DataFrame(results).T
df_results
```
